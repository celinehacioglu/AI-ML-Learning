```python
import numpy as np
import math

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

total = v1 + v2
print("v1 + v2 =", total)

difference = v2 - v1
print("v2 - v1 =", difference)

v3 = v1 * 2
print("v1 * 2 =", v3)

dot = np.dot(v1, v2)
print("v1 · v2 =", dot)

print("manual calculation: 1*4 + 2*5 + 3*6 =", 1*4 + 2*5 + 3*6)

length = np.linalg.norm(v1)
print("v1 length:", length)

manuel_length = math.sqrt(1**2 + 2**2 + 3**2)
print("Manual length:", manuel_length)


eleman_multiplication = v1 * v2  # This is not dot product!
print("Multiplication by element:", eleman_multiplication)

# ============================================


v = np.array([3, 4])
length_v = np.linalg.norm(v)
print("Length of v:", length_v)

dot = v1 @ v2
print("v1 . v2 =", dot)

v1 = [10, 20, 30] 
v1_normalized = v1 / np.linalg.norm(v1)
print("Normalized v1:", v1_normalized)

# cos(θ) = (v1 · v2) / (||v1|| * ||v2||)
cos_theta = v1 @ v2 / (np.linalg.norm(v1) * np.linalg.norm(v2))
print("Cosine of the angle between v1 and v2:", cos_theta)

```

```python
import numpy as np

m1 = np.array([[1, 2], 
               [3, 4]])
print("\n1. Matris m1:")
print(m1)
print("   Boyut:", m1.shape)

# zero matrisi
zeros_m = np.zeros((3, 3))
print("\n2. Zero Matrix (3x3):")
print(zeros_m)

#  Units matrisi
ones_m = np.ones((2, 3))
print("\n3. Units Matrix (2x3):")
print(ones_m)

# identity matrix
identity = np.eye(3)
print("\n4. Identity Matrix (3x3):")
print(identity)

# Random matris
random_m = np.random.rand(2, 2)
print("\n5. Random Matrix (2x2):")
print(random_m)

```

```python

m1 = np.array([[1, 2], 
               [3, 4]])

m2 = np.array([[5, 6], 
               [7, 8]])
print("\nm2 Matrix:")
print(m2)

addition_m = m1 + m2
print("\n6. Matrix Addition (m1 + m2):")
print(addition_m)
print("   Manuel: [1+5, 2+6] = [6, 8]")
print("           [3+7, 4+8] = [10, 12]")

m3 = m1 * 2
print("\n7. Scalar Multiplication (m1 * 2):")
print(m3)
print("   Manuel: [1*2, 2*2] = [2, 4]")
print("           [3*2, 4*2] = [6, 8]")


matrix_multiplication = m1 @ m2
print("\n8. Matrix Multiplication (m1 @ m2):")
print(matrix_multiplication)
print("   Manuel: [(1*5)+(2*7), (1*6)+(2*8)] = [19, 22]")
print("           [(3*5)+(4*7), (3*6)+(4*8)] = [43, 50]")


multiplication = m1 * m2
print("\n9. Element-wise Multiplication (m1 * m2):")
print(multiplication)
print("   Manuel: [1*5, 2*6] = [5, 12]")
print("           [3*7, 4*8] = [21, 32]")

m1_T = m1.T
print("\n10. Transpose (m1.T):")
print("    Original m1:")
print("   ", m1)
print("    Transpose:")
print("   ", m1_T)
print("    (Rows ↔ Columns)")

# Matrix-Vector çarpımı
v = np.array([1, 2])
result = m1 @ v
print("\n11. Matrix-Vector Multiplication (m1 @ v):")
print("    m1:")
print("   ", m1)
print("    v:", v)
print("    Result:", result)
print("    Manual: [(1*1)+(2*2), (3*1)+(4*2)] = [5, 11]")

```
